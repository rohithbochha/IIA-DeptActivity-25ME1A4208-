```python
crop_data={
  "crop_name":"Rice",
  "scientific_name":"Oryza sativa",
  "crop_type":"Cereal",
  "season":"Kharif",
  "ideal_temperature_celsius":25,
  "annual_rainfall_mm":"1200",
  "soil_type":"Clavey loam",
  "ideal_soil_ph":"6.5",
  "growing_duration_days":"120",
  "water_requirement":"High",
  "average_yield_per_hectare_kg":4000,
  "primary_nutrients":["Nitrogen","phosphorus","Pottasium"],
  "common_pests":["Stem Borer","Brown","Planthopper"],
  "harvesting_method":"Manual/Mechanical",
  "storage_humidity_percent":13
}
print(crop_data)
crop_data["Area"]="Eluru"
crop_data["Pesticide"]="Urea"

print(crop_data)
```

```python
import pandas as pd
from google.colab import files

# Prepare the dictionary for DataFrame conversion
# Non-scalar values like lists need to be formatted or kept as lists depending on user preference.
# We will convert list values to comma-separated strings for cleaner CSV representation.
processed_data = {}
for k, v in crop_data.items():
    if isinstance(v, list):
        processed_data[k] = ", ".join(v)
    else:
        processed_data[k] = v

# Create a DataFrame with one row
df = pd.DataFrame([processed_data])

# Save to CSV
csv_filename = 'crop_data.csv'
df.to_csv(csv_filename, index=False)

# Display DataFrame and trigger download
display(df)
files.download(csv_filename)
```
